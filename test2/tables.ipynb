```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned CSV
data = pd.read_csv("data/messy_data_cleaned.csv")
data["people_served"] = pd.to_numeric(data["people_served"])

# table
chart_data = data.pivot_table(
    index="disability_category",
    columns="program_status",
    values="people_served",
    aggfunc="sum",
    fill_value=0,
)

# plot
chart_data.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6),
    color={
        "Active": "#cea213",
        "Inactive": "#0b1aa1",
        "Pending": "#7cf159",
    },
    edgecolor="black",
)

plt.title("People Served by Disability Category")
plt.xlabel("Disability Category")
plt.ylabel("People Served")
plt.xticks(rotation=0)
plt.legend(title="Program Status")
plt.tight_layout()
plt.show()
```
